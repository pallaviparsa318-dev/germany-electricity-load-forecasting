# Germany electricity load: short EDA
Synthetic sample; timestamps are UTC.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("../data/sample_data.csv", parse_dates=["timestamp"])
print(df.isna().sum())
display(df.describe())
df.set_index("timestamp")["load_mw"].iloc[:24*14].plot(figsize=(12, 3), title="Two weeks of hourly load")
plt.show()
```

```python
df.groupby(df["timestamp"].dt.hour)["load_mw"].mean().plot(title="Average UTC hour profile")
plt.show()
```
